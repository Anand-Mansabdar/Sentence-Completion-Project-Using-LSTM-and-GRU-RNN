```python
import string
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
df = pd.read_csv("quotes.csv")
df.head()
```

```python
df['quote'][1]
```

```python
df.shape
```

```python
quotes = df['quote']
print(quotes.head())
```

```python
# Data preprocessing
quotes = quotes.str.lower()
quotes.head()
```

```python
# To remove punctuations  

translator = str.maketrans("", "", string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))
```

```python
quotes.head()
```

```python
# Tokenization
vocab_size = 10000

tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index = tokenizer.word_index
print(len(word_index))

list(word_index.items())[:10]  # Prints all the unique words in the dataset and assigns numerical values to each word based their frequency

# 'the' is assigned 1 as its the most frequently occuring word
```

```python
sequence = tokenizer.texts_to_sequences(quotes)
print(quotes[0])
print(sequence[0])
```

```python
for i in range(3):
  print(f"The {i+1} quote is:",quotes[i])
  print(f"The tokenized sequence of the quote {i+1}is:",sequence[i])
```
