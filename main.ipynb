```python
import string
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
```

```python
df = pd.read_csv("quotes.csv")
df.head()
```

```python
df['quote'][1]
```

```python
df.shape
```

```python
quotes = df['quote']
print(quotes.head())
```

```python
# Data preprocessing
quotes = quotes.str.lower()
quotes.head()
```

```python
# To remove punctuations  

translator = str.maketrans("", "", string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))
```

```python
quotes.head()
```

```python
# Tokenization
vocab_size = 10000

tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index = tokenizer.word_index
print(len(word_index))

list(word_index.items())[:10]  # Prints all the unique words in the dataset and assigns numerical values to each word based their frequency

# 'the' is assigned 1 as its the most frequently occuring word
```

```python
sequence = tokenizer.texts_to_sequences(quotes)
print(quotes[0])
print(sequence[0])
```

```python
for i in range(3):
  print(f"The {i+1} quote is:",quotes[i])
  print(f"The tokenized sequence of the quote {i+1}is:",sequence[i])
```

```python
# Input and Output features
X = []
y = []

for seq in sequence:
  for i in range(1, len(seq)):
    input_seq = seq[:i]
    output_seq = seq[i]
    X.append(input_seq)
    y.append(output_seq)
```

```python
print(len(X))
print(len(y))
```

```python
# Applying padding so that we have the values in X in same length.

# Before Padding -> After Padding
# [128]  -> [0, 0, 128]
# [136, 128] -> [0, 136, 128]
# [12, 136, 128] -> [12, 136, 128]

max_length = max(len(x) for x in X)
print(max_length)
```

```python
X_padded = pad_sequences(X, maxlen=max_length, padding="pre")
X_padded[0]
```

```python
y = np.array(y)
print(X_padded.shape)
print(y.shape)
```

```python
# One hot encoding the output(y)
y_encoded = to_categorical(y, num_classes=vocab_size)

print(y.shape)
print(y_encoded.shape)
```

```python
y_encoded[0]
```

```python
# Creating models
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Embedding, LSTM, GRU, Dense

embed_dim = 50
rnn_units = 128

rnn_model = Sequential()
rnn_model.add(Embedding(input_dim=vocab_size, output_dim=embed_dim, input_length=max_length))

rnn_model.add(SimpleRNN(units=rnn_units))
rnn_model.add(Dense(units=vocab_size, activation="softmax")) # Output layer
```

```python
rnn_model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])
```

```python
rnn_model.summary()
```

```python
lstm_model = Sequential()

lstm_model.add(Embedding(input_dim=vocab_size, output_dim=embed_dim, input_length=max_length))
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=vocab_size, activation="softmax"))
```

```python
lstm_model.compile(optimizer="adam", loss="categorical_crossentropy", metrics=["accuracy"])

lstm_model.summary()
```
